```python
import logging

import dotenv

logging.basicConfig(level=logging.DEBUG)
dotenv.load_dotenv()
import newaresql
import newaresql.local
```

```python
class CallBack:
    def __init__(self, number_of_tests: int = 10, limit_days: int = 7):
        # Adjust the total number of tests to process
        self.counter = number_of_tests
        self.limit_days = limit_days  # Maximum allowed duration in days

    def __call__(self, test) -> bool:
        if test["end_time"] is None:
            return False

        duration_days = (test["end_time"] - test["start_time"]).days
        if duration_days > self.limit_days:
            return False

        if self.counter > 0:
            self.counter -= 1
            return True
        return False
```

```python
with newaresql.connect() as connector:
    callback = CallBack(limit_days=7, number_of_tests=10)
    newaresql.local.update(callback=callback, connector=connector, workers=8)
```
